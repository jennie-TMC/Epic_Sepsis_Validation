```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
folder_path = 'C:/Users/jenan/OneDrive - Tufts Medicine/code/EDSValidation_Jennie/Data/Time_Dependent_All/No Refractory/Threshold_25_Lookforward_12'
results_df = pd.read_csv(os.path.join(folder_path, 'Time_dependent_metric_results.csv'))
```

```python
# subset df for 4 days, set legend to be in days
subset_results_df = results_df[results_df['Time Point'] <= 384]
subset_results_df['Time Point Days'] = subset_results_df['Time Point'] / 96
subset_results_df['Prevalence'] = subset_results_df['Prevalence'] * 100
```

```python
# plot (4 days, x axis in days)
fig, axs = plt.subplots(10, 1, sharex=True, figsize=(10, 20))

axs[0].plot(subset_results_df['Time Point Days'], subset_results_df['Prevalence'], label='Prevalence')
axs[0].set_ylabel('Prevalence')
axs[0].set_title('Prevalence calculated at each hour, average: ' + str(round(np.mean(subset_results_df['Prevalence']), 2)))
axs[0].legend()

axs[1].plot(subset_results_df['Time Point Days'], subset_results_df['AUC'], label='AUC')
axs[1].set_ylabel('AUC')
axs[1].set_title('AUC calculated at each hour, average: ' + str(round(np.mean(subset_results_df['AUC']), 2)))
axs[1].legend()

axs[2].plot(subset_results_df['Time Point Days'], subset_results_df['Sensitivity'], label='Sensitivity')
axs[2].set_ylabel('Sensitivity')
axs[2].set_title('Sensitivity calculated at each hour, average: ' + str(round(np.mean(subset_results_df['Sensitivity']), 2)))
axs[2].legend()

axs[3].plot(subset_results_df['Time Point Days'], subset_results_df['Specificity'], label='Specificity')
axs[3].set_ylabel('Specificity')
axs[3].set_title('Specificity calculated at each hour, average: ' + str(round(np.mean(subset_results_df['Specificity']), 2)))
axs[3].legend()

axs[4].plot(subset_results_df['Time Point Days'], subset_results_df['PPV'], label='PPV')
axs[4].set_ylabel('PPV')
axs[4].set_title('PPV calculated at each hour, average: ' + str(round(np.mean(subset_results_df['PPV']), 2)))
axs[4].legend()

axs[5].plot(subset_results_df['Time Point Days'], subset_results_df['NPV'], label='NPV')
axs[5].set_ylabel('NPV')
axs[5].set_title('NPV calculated at each hour, average: ' + str(round(np.mean(subset_results_df['NPV']), 2)))
axs[5].legend()

axs[6].plot(subset_results_df['Time Point Days'], subset_results_df['TP'], label='TP')
axs[6].set_ylabel('TP')
axs[6].set_title('TP calculated at each hour')
axs[6].legend()

axs[7].plot(subset_results_df['Time Point Days'], subset_results_df['FP'], label='FP')
axs[7].set_ylabel('FP')
axs[7].set_title('FP calculated at each hour')
axs[7].legend()

axs[8].plot(subset_results_df['Time Point Days'], subset_results_df['TN'], label='TN')
axs[8].set_ylabel('TN')
axs[8].set_title('TN calculated at each hour')
axs[8].legend()

axs[9].plot(subset_results_df['Time Point Days'], subset_results_df['FN'], label='FN')
axs[9].set_ylabel('FN')
axs[9].set_title('FN calculated at each hour')
axs[9].legend()

axs[9].set_xlabel('Time Point Days')

# Adjust spacing between subplots
plt.tight_layout()

# save plot
#plt.savefig(os.path.join(folder_path, 'metric_figure_improved.png'))

# Show plot
plt.show()
```
